### Operations on variables
You can then use this variable to perform different operations on 

First set your variable

```python
myVar = 2 + 3
print(myVar)
```

Now we can use it

```python
print(myVar+3)               # adding
print(myVar*2)               # multiplying
print(myVar/myVar)           # dividing 
print(myVar**2)              # squaring (order of 2)
print(np.sqrt(myVar))        # square rooting         
print(np.log(myVar))         # ln of a value
print(np.log10(myVar))       # log to base 10
```

Python uses the order of BODMAS (brackets, order(power), divide, multiple, addition and subtraction).

### Trigonometry

The value of pi is built into _numpy_. Trigonometric functions use radians. Degrees need to be converted to radians either 90/360*2*np.pi or using the _numpy_ function np.deg2rad().  

```python
x=np.pi                # assigns pi to x
print(x)
print(np.cos(x))              # note uses radians
print(np.sin(x))
print(np.tan(x))              
print(np.arctan(x))           # inverse tan   
print(np.cos(np.deg2rad(90))) # cos assuming x is in degrees
```

### Limitations of Python

Python and most other programming languages have limitations regarding the largest/smallest number and precision. 

```python
y=1e307                     # a large number
print(y, y*100)             # too large for Python (inf is infinity)
y=1e-322                    # a small number 
print(y, y/100)             # too small for Python to read (so reads 0)
print(np.finfo(float).eps)  # the smallest number Python can 'see' between two values
```

### Multiple entries on one line
As the semi-colon tells Python that you have completed the line, you can put multiple entries on one line to save space.

```python
a = 5; b = 4
print(a,b)
c = a + b
print(c)
```

Alternatively you can assign values to multiple variables in one line like this 

```python
a, b = 5, 8
print(a,b)
c = a + b
print(c)
```
