# Tutorial: Functions in Python

## Introduction to Functions

In Python, a function is a reusable block of code that performs a specific task. In research, we use functions to automate repetitive calculations, like converting units or calculating material properties.

### The Syntax

A functions is defined with the `def` keyword:

```
def function_name(parameter):
    """
    Comment to describe the function and their input and outputs.
    More complex functions also include examples and how to use it.
    """
    result = parameter * 2 # some statements, operations, calculations, ...
    return result          # returns result to main program.

# Calling the function
a = function_name(5)
# Prints results
print(a)
```

`return` is optional and is used, if you want the function to give a value back to the main program so you can use it.

- **Parameters**: The variables listed in the function definition (`parameter`).

- **Arguments**: The actual values passed into the function when you call it (`5`).

```
def greet_user(username):
    """
    Display a simple greeting.
    username: string value
    """
    print(f'Hello, {username}!')

# Calling the function
name = 'Joachim'  # string
greet_user(name)
```

In the example above, no value is return to the main program. Any variables defined inside the function are not available to the main program, e.g. there is no variable called `username` accessible in the main program.

Function parameter is `username`. Argument to the function is the variable `name`.

### Default values

If you provide too many or too few arguments to your function, you will receive an error message. However, a function parameter can have a default value, which is used if no argument is provided for that parameter when the function is called. This allows the function to be called with fewer arguments than it is defined to accept, making those parameters optional.

```
def add_numbers(number, addend = 3):
  """
  Adds 2 numbers
  number: float or int
  addend: float or int, optional input: default is 3
  """
  result = number + addend
  return result
```
