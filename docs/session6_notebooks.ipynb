# Tutorial: Reading in and writing out data

This live script will show you how to read in a file of data, how to process, fit the data, and then print the answer to another file. 

## Reading in from a file
You will need to first download the 'test.txt' file from BlackBoard and upload it to your workspace on Colab ([see overview page](#session6)).

_Numpy_ provides the `np.loadtxt()` function to read in a file. To to so you need to specify three things:

- The name of the file. The file needs to be in your workspace and not locally on your Desktop or Laptop.
- `delimiter`: This is the symbol that separates the values in the file. This could be a space whitespace including tab (default), a comma ',', '|' or some other symbol. The file separated with commas are more commonly known as a CSV file, or a Comma Separated Variable file. 
- `comments`: The number of comments lines at the start of the file. In this case, if you open the 'test.txt' in word pad of equivalent or `!cat test.txt` in the code cell, you will see 2 rows of text headings. We must make sure Python reads this in as text and not numerical values. So we have to specify this.
