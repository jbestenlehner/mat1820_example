# Tutorial: Fitting data with uncertainties on both axes

So far we have only considered errors in the dependent variable (y-axis). We used a standard least square regression. If we have uncertainties (errors) in both independent (x-axis) and dependent (y-axis) variable, we will need to use a method called weighted **Orthogonal Distance Regression (ODR)** or **Total Least Squares**. ODR minimises the perpendicular (orthogonal) distance from the data points to the fitted curve.

```python

              weight_x=1/variance_x, weight_y=1/variance_y)
```

- `f` is your model function (e.g. Ohm's law).
- `beta0` is the initial guess of your parameter to fit. For example, you can use the result for Task 1 or Task 2. For simple and linear models like Ohm's law, the initial guess does not need to be precise but is crucial in complex non-linear models. 
- `xdata` and `ydata` are your measurements or data set.
- `weight_x` and `weight_y` are your weights for your data set and are the inverse of the squared standard deviations (variance), e.g. **`weight_x = 1/sigma_x^2` and `weight_y = 1/sigma_y^2`.**
