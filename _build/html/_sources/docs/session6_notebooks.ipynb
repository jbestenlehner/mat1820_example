## Reading in from a file
You will need to first download the 'test.txt' file from BlackBoard and upload it to your workspace on Colab ([see overview page](#session6)).

_Numpy_ provides the `np.loadtxt()` function to read in a file. To to so you need to specify at least the following things:

- The name of the file. The file needs to be in your workspace and not locally on your Desktop or Laptop.
- `delimiter`: This is the symbol that separates the values in the file. This could be a space whitespace including tab (default), a comma ',', '|' or some other symbol. The file separated with commas are more commonly known as a CSV file, or a Comma Separated Variable file. 
- `comments`: The number of comments lines at the start of the file. In this case, if you open the 'test.txt' in word pad of equivalent or `!cat test.txt` in the code cell, you will see 2 rows of text headings. We must make sure Python reads this in as text and not numerical values, but the text is not specifically indicated as a comment with a symbol at the beginning of the line like '#' or '!'. So we have to specify this with the `skiprows` parameter.
- `skiprows`: If text is not specified as comment, we must make sure Python does not read this in as text. So we have to tell how many lines to skip at the beginning of the row.
- `unpack`: Default is `False`. All columns are read into a single 2D `np.array()`. If you want to read each column in a separate 1D array, set `unpack` to `True`. 

First we need to check the content of the file to see, if there are headers/comments present, what is the delimiter and how the data are structured in the file. To do so you can either open the file with an text editor like word pad or execute the following command in your code cell.

```python
!cat  test.txt # prints the content of the file to the screen
```

We notice, that there are two lines of comments/header, the delimiter is whitespace and there are two columns of data (x and y).

Note: If you open the file in Excel, Excel will read it into columns and you won't be able to see, what the delimiter is.

```python
import numpy as np
import matplotlib.pyplot as plt

xy_data = np.loadtxt('test.txt', skiprows = 2)  # Note: delimiter is white space and does not need to be specified (default).
                                                 # header is not indicated by a symbol. So we use skiprows to skip the first 2 lines.
print(xy_data.shape)                             # prints shape
x_data, y_data = np.loadtxt('test.txt', skiprows = 2, unpack=True) # reads each column into a separated array
print(x_data.shape, y_data.shape)

plt.plot(xy_data[:,0], xy_data[:,1],'r-')                           # plot the data
plt.plot(x_data, y_data,'b--')
plt.show()
plt.close()
```

## Read in structure from file

To read in the file structure in a `DataFrame` you can use the `pd.read_csv()` function from Pandas.

Note: as it is a text file separated by whitespaces we need to set the parameter `delim_whitespace = True`. 

The first line of the header/comment describes the data file ('test file') and we can ignore it. The second row provides us information about the columns, which would be sensible labels for the pandas `DataFrame`. So we only skip the first row.

```python
import pandas as pd

df_xy = pd.read_csv('test.txt', delim_whitespace = True, skiprows=1) # reads data into Pandas dataframe.
print(df_xy)          # prints content of dataframe
print(df_xy.columns)  # shows column labels
print(df_xy.dtypes)   # prints data type for each column

plt.plot(df_xy['x'], df_xy['y'],'r-') # plot data
plt.plot(df_xy.x, df_xy.y,'b--')      # if labels do not contain spaces or special signs like '\', '|', '/', '#', ..., you can access columns with '.<column label>.
plt.show()
plt.close()
```

## Writing to a file 

We use Python's built in functions `open()` and `write()`. To write to a file you have to first open a file with the permission to write to it, given by the option 'w'. You then use `write()` with the string-formatter `.format()` to write to the file ([See also session 4](#session4)). Finally close the file.

```python
fid = open('test_write.txt', 'w')   # open file with write permission. Overwrites any existing file called 'test_write.txt'
fid.write('heading\n')              # Writes some header. Note: '\n' new lines needs to be added to the end of the string.
fid.write('{:.0f} {:.2f}\n'.format(2, np.random.rand())) # formats number values and writes them into a file. 
fid.close()                         # close the file
```

- The fid variable stores the open file. 
- The `open()` functions takes in two parameters. The name of the file and the permission to write to the file (See overview page of this session).
- The `write()` function has one input which is a string. Therefore, we need format our number values to a single string. 
- The first `write()` expression writes a line of text at the top. 
- The second `write()` expression writes two values, the value of 2 followed by a random number, separated by a space.
- It is really important you then close the file before the algorithm ends. Not all data might be written to the file.
