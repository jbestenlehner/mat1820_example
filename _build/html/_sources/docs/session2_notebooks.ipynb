# Tutorial: Vectors and matrices

In _NumPy_ vector is 1 dimenisonal array where you can store information like data. 

A matrix is a 2 dimenisonal array like a table. It has both rows and columns. In principle you can have multi-dimensional arrays, but we will deal with this later.

To stor numbers in an array to a variable, you can do this in a few ways.

````python
import numpy as np

array_1D = np.array([1, 2, 3, 4, 5])          # generates an 1D array
print(array_1D)
array_2D = np.array([[1, 2, 3], [4, 5, 6]])   # generates a 2x3 2D array
print(array_2D)
array_2D_transpose = array_2D.T               # note ".T" for transpose
print(array_2D_transpose)
array_1D_zero = np.zeros(5)                   # generates an 1D array filled with zeros
print(array_1D_zero)
array_2D_one  = np.ones((3,2))                # generates 3x2 2D array filled with ones, note the "(...)", which is the shape of the array.
print(array_2D_one)
array_1D_random = np.random.rand((3))         # generates 3 random numbers in a 1D array
array_3D_random = np.random.rand((3,3,3))     # generates 27 random numbers in a 3D array
````

Note: _np.random.rand_ creates an array of the given shape and populate it with random samples from a uniform distribution over **[0, 1)** (excluding 1).

If you would like to know more _numpy.random_ package, you can have a look [here](https://numpy.org/doc/2.2/reference/random/index.html#module-numpy.random).

To generate evenly spaced values within a given interval the function _np.arange_ is used.

````python
print(np.arange(10))       # generates an array from 0 to 9. Note in Python the default start value is zero and not one.
print(np.arange(1,11))     # of course you can start from one, but then start and end of interval must be provided. Note that the end value is not included (Values are generated within the half-open interval).
print(np.arange(1,11,0.5)) # step size are defined with the third value.
a = np.arange(10) + 1      # generates array from 1 to 10, because we add 1 to the array.
a = a.reshape(5,2)         # reshapes the array to a 2D array with 5 rows and 2 columns.
print(a)
````

A quick quick way of generating an array is using a function called _np.linspace_. Return evenly spaced numbers over a specified interval (linear). 

````python
print(np.linspace(1,5,5))      # first number is the starting value of the sequence, second number is the end value of the sequence and third number is the number of samples to generate. If no third number is given the default is 50. While the first 2 numbers can have any value the third number must be non-negative.
print(np.linspace(5,-5))       # descending sequence.
````

Note: Evenly spaced numbers in log space can be generated with _np.logspace_.

### Vector operations

In Python vectors (1D), matrices (2D) and tensors (N-D) are all called arrays. They support multidimensional array algebra, that vectors, matrices and tensors are used for.

Data are often stored in 1D arrays (vectors), e.g. [Task 3](#task-3). You might expect that those data are either stored in a column vector (**Nx1**) or row vector (**1xN**). By default in Python the 1D array is neither **Nx1** nor **1xN**. It is just **N**. Therefore, a _transpose_ on 1D array does nothing. It first this might be a bit confusing, but it comes with many advantages.

````python
a = np.arange(1,4)         # generates an array with [1,2,3]
print(a.shape)             # prints the shape of the array
a_column = a.reshape(3,1)  # reshapes a into column vector
print(a_column)            # prints column vector
print(a_column.shape)      # prints the shape of the array, note the difference to array a
print(a_row)               # prints row vector
a_row = a.reshape(1,3)     # reshapes a into row vector
````

#### Vector multiplications

In Python multiplication operation are defined as follow: ``*`` means element-wise multiplication, while ``@`` means matrix multiplication. The differences between ``*`` and ``@`` become more obvious when dealing with multi-dimensional arrays.

Consider the following element-wise multiplication:

```{math}
\begin{bmatrix}
1 & 2 & 3\\
\end{bmatrix}
\begin{bmatrix}
1 & 2 & 3\\
\end{bmatrix}
=
\begin{bmatrix}
1 & 4 & 9\\
\end{bmatrix}
```

or

```{math}
\begin{bmatrix}
1 \\
2 \\
3 \\
\end{bmatrix}
\begin{bmatrix}
1 \\
2 \\
3 \\
\end{bmatrix}
=
\begin{bmatrix}
1 \\
4 \\
9 \\
\end{bmatrix}

````python
print(a*a)               # gives an array [1, 4, 9]
print(a_row*a_row)       # gives an row array
print(a_column*a_column) # gives a column array
print(a_row@a_row)       # gives an error message
print(a_column@a_column) # gives an error message
````

Let's see what happens when we replace a row with a column vector/array and vice versa.

````python
print(a_row*a_column)   # gives a 3x3 matrix/array
#while
print(a_row@a_column)   # gives a scalar
#or
print(a_column*a_row)   # gives a 3x3 matrix/array
#while
print(a_column@a_row)   # gives a 3x3 matrix/array
````

This is all a bit confusing and the results might not be the result, you might have expected.

There are 2 vector multiplication present:

1. row vector times column vector:

```{math}
\begin{bmatrix}
1 & 2 & 3\\
\end{bmatrix}
\begin{bmatrix}
1 \\
2 \\
3 \\
\end{bmatrix}
=
14
```

This also called a dot product (row times column). See also overview page of this session 2. 

2. column vector times row vector:

```{math}
\begin{bmatrix}
1 \\
2 \\
3 \\
\end{bmatrix}
\begin{bmatrix}
1 & 2 & 3\\
\end{bmatrix}
=
\begin{bmatrix}
1& 2& 3\\
2& 4& 6\\
3& 6& 9\\
\end{bmatrix}
```

To avoid mistakes and save you having to type a lot of transposes Python defines by default a one-dimensional array of shape N.

````python
# row times column vector
np.inner(a,a)  # gives a scalar
#or
a@a            # by default Python treats the vector of shape N before the @ as a row vector and after the @ as a column vector.
# column times row vector
np.outer(a,a)  # gives a 3x3 matrix/array
````

In the case of matrices (2D arrays) or N-dimension arrays the differences between between ``*`` and ``@`` are clearer.

````python
m = np.arange(1,5).reshape(2,2) # generates a 2x2 array with values from 1 to 4.
print(m)
print(m*m)                      # element wise multiplication
print(m@m)                      # matrix multiplication, see also overview page for session 2.
````

## Matrices

A matrix is a way of storing information using both rows and columns (two dimensions).

### Creating a matrix: part 1
You first give number of rows (they go across), then columns (which go down). You can set up matrices in a number of ways. The first to to generate a matrix using quick functions.

````python
m = 3                      # set up a value of m to be 3
n = 4                      # set up a value of n to be 4
print(np.zeros(n))         # create a vector of 4 zeros
print(np.eye(m,n))         # create a matrix of 3x4 with 1's down the diagonal (an identity matrix)        
print(np.ones(n,m))        # create a matrix of 4x3 of ones.
````

### Creating a matrix: part 2

The second way is define the values. This could be through combining arrays invidually and putting them together.

````python
array_1 = np.array([1,2,3,4])              # set an array 1
array_2 = np.array([3,4,5,6])              # set an array 2
A = np.column_stack((array_1, array_2))    # combine into a 4x2 matrix
print(A)
B = np.vstack((array_1, array_2))          # combine into a 2x4 matrix
print(B)
#or
B_1 = np.array([array_1,array_2])
print(B_1)
#or in one line
B_2 = np.array([1,2,3,4], [3,4,5,6]) 
print(B_2)
C =  np.concatenate((array_1, array_2))    # concatenate into a 1D array
````
