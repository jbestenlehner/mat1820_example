### Basic line graph

We use the _plt.plot()_ function to plot data. After we finished our plot instruction, we create the graph with _plt.show()_. With _plt.close()_ you clean up the current plot environment and free up memory space. This becomes important, if you plot large datasets or hundreds/thousands of plots at once.

```python
A = np.arange(1,6)    # Some data in array A
B = 2*A               # Some data in array B
plt.plot(A,B)         # Plot data A as x values and B as y values
plt.plot(B,A)         # Plot data B as x values and A as y values
plt.show()            # Shows the plot/graph
plt.close()           # Closes the plotting environment and frees up memory.
```

No we can use our vector knowledge to create data such as this.

```python
x = np.linspace( 0, 2*np.pi, 100)
y = np.sin(x)
plt.plot(x,y)
plt.show()
plt.close()
```

## Matrices and plotting
We are not restricted to plotting different arrays but could use two rows or columns from the same array. Be careful with the size of the array.  We can also plot multiple data sets in the same graph until we create the graph with _plt.show()_.

```python
C_1 = np.array([2,4,6,8,10]) # array 1
C_2 = -C_1                   # array 2
C = np.vstack((C_1, C_2))    # Stack them along the vertical axis
plt.plot(C[0,:], C[1,:])     # Plot all of row 1 against row 2
plt.plot(C[1,:], C[0,:])     # Plot all of row 2 against row 1
plt.plot(A, B)               # Plots data from above
plt.show()                   # Note: the graph is only created when the function plt.show() is executed.
plt.close()
```

## Bar graph
A bar chart is really straightforward to generate too:

```python
y = np.array([75, 91, 105, 123.5, 131, 150, 179, 203, 226, 249, 281.5])
x = np.arange(y.size)+1 #creates an array from 1 to the size of y.
plt.bar(x,y) #x gives the position of the bars while y defines the height of the bars. Change it to see the effect.
plt.show()
plt.close()
```

You can use multiple arrays to represent different data sets, too.

You can also add categories to the bars as in this example.

```python
categories = ('students', 'lecturers', 'professors')
y = np.array([5, 10, 15])
plt.bar(categories, y)
plt.show()
plt.close()
```

## Formatting of plots

There are lots of ways of control the format of a graph that we plot.

```python
A = np.array([1,2,3,4,5])
B = np.array([1,8,27,64,125])

plt.plot(A,B,'r')         # We can add format commands with characters in inverted commas such as the colour
plt.plot(A,B+10,'x')      # The marker for the data
plt.plot(A,B+20,':')      # The type of line
plt.plot(A,B+30,'bs')     # Or together
plt.show()
plt.close()
```

See the second part of the overview page on _matplotlib_ or use the help function to see and learn all the basics terms.

We can add `labels`, `limits` and `legend` to our graphs for the axis.

```python
plt.plot(A,B, label='Data AB') # adds label to the plot, which is shown in the legend
plt.xlabel('A')      # Labels x axis
plt.ylabel('B')      # Labels y axis
plt.ylim((-25,150))  # Sets the plotting range for the y axis
plt.xlim((5,1))      # Inverts the x axis and sets the plotting range for the x axis
plt.legend()         # adds legend
plt.show()
plt.close()
```

You can save the graph as a png file with a right click on the plot and select "Save image as ...". Later in the course we will learn, how to save plots automatically, when we will do more advance and sophisticated graphs and figures.

Sometimes we want to plot multiple sets of data but they do not fit well on the same axis.

```python
A = np.array([1,2,3,4,5])
B = np.array([1,8,27,64,125])
C = np.array([1150,1280,1315,1360,1408])
plt.plot(A,B)
plt.plot(A,C)
plt.show()
plt.close()
```

You can share axis for particular plots with _plot.twinx()_ (shared x-axis) or _plot.twiny()_ (shared y-axis).

```python
plt.plot(A,B, 'r+')                       # plots data with red plusses
plt.xlabel('A')
plt.ylabel('B', color='r')                # Labels left y-axis with red
plt.tick_params(axis='y', labelcolor='r') # Change colour of y-axis tick labels
plt.twinx()                               # instantiate a second Axes that shares the same x-axis
plt.plot(A,C,'bo')
plt.ylabel('C', color='b')                # Labels right y-axis with blue
plt.tick_params(axis='y', labelcolor='b') # Change colour of y-axis tick labels
plt.show()
plt.close()
```

Note: we changed the colour of the axis labels, so that we know, which data points belong to which y-axis.

## Two plots in one - subplots
Instead of plotting the graphs over one another you can put them as separate items in the same figure. A subplot creates an array of figures. 

```python
x = np.arange(0,50,0.5)     # generates x values
y = 5*x**2                   # generates y values
plt.subplot(1,3,1)          # sets first graph in a 3x1 system as plot 1
plt.plot(x,y)               # line plot of x y 
plt.subplot(1,3,2)          # sets second graph in a 3x1 system as plot 2
plt.loglog(x,y)             # log axes (both) for x and y to the base 10 (default)
plt.subplot(1,3,3)          # sets third graph in a 3x1 system as plot 3
plt.yscale('log', base=5)   # Set y axis to base 5
plt.xscale('log', base=2)   # Set x axis to base 2
plt.plot(x,y)
plt.show()
plt.close()
```

## Histograms 
Histograms are really useful in engineering and Python makes it very easy to generate them.

```python
a = np.random.rand(100) - np.random.rand(100) # create a set of 100 random numbers
plt.subplot(2,2,1)                     # sets first graph in a 2x2 system as plot 1
plt.hist(a)                            # number of bins is 10 (default)
plt.subplot(2,2,2)                     # sets second graph in a 2x2 system as plot 2
plt.hist(a, bins=15)                   # plot with 15 bins
plt.subplot(2,2,3)                     # sets third graph in a 2x2 system as plot 3
plt.hist(a, bins=np.arange(-1,1.5,0.5))# Plot them as a histogram with 0.5 bins from -1 to 1
plt.subplot(2,2,4)                     # sets forth graph in a 2x2 system as plot 4
plt.hist(a, bins=np.arange(-1,1.1,0.1))# Plot them as a histogram with 0.1 bins from -1 to 1
plt.show()
plt.close()
```
