## Defining, assigning and manipulating variables

Variables are used for storing data that can be referenced and manipulated throughout a programme. A variable is essentially a name that is assigned to a value. To assign a value to a variable the assignment operator `=` is used. This is not a mathematical operation. 

When defining variables, make sure you do not use variables with spaces, or variables that start with a number. Avoid key words like sin, cos, exp, mean, print etc.

```python
a = 2
b = 3
a + b
```

After executing the code cell above you have defined the variables `a` and `b` with the values `2` and `3`, respectively. If you click on _{} variables_ icon at the bottom left, a tab on your right hand side will open, which shows the properties of your variables.

### Trigonometry

The value of pi is built into _numpy_. Trigonometric functions use radians. Degrees need to be converted to radians either `90/360*2*np.pi` or using the _numpy_ function `np.deg2rad()`.  

```python
x = np.pi                    # assigns pi to x
print(x)
print(np.cos(x))             # note uses radians
print(np.sin(x))
print(np.tan(x))              
print(np.arctan(x))          # inverse tan
x = 90   
print(np.cos(np.deg2rad(x))) # cos assuming x is in degrees
```

### Limitations of Python

Python and most other programming languages have limitations regarding the largest/smallest number and precision. 

```python
y = 1e307                   # a large number
print(y, y*100)             # too large for Python (inf is infinity)
y = 1e-322                  # a small number 
print(y, y/100)             # too small for Python to read (so reads 0)
print(np.finfo(float).eps)  # the smallest number Python can 'see' between two values
```

### Multiple entries on one line
As the semi-colon tells Python that you have completed the line, you can put multiple entries on one line to save space.

```python
a = 5; b = 4
print(a,b)
c = a + b
print(c)
```

Alternatively you can assign values to multiple variables in one line like this 

```python
a, b = 5, 8
print(a,b)
c = a + b
print(c)
```
