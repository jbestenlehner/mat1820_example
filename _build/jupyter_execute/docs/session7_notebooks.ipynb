# Tutorial: Data fitting

Once a data set has been read in, one common thing is to fit it for analysis. This could be to remove noise or calculate key parameters such as gradients and intercepts. Here we shall look at the spline fitting and least squares fitting. 

![fitting data](https://raw.githubusercontent.com/jbestenlehner/mat1820_example/refs/heads/main/pictures/session7_spline_fitting.png)

**Spine fitting:**

- Also known as spline interpolation. It is a way of interpolating between data points but does not provide any numerical understanding – just joins the dots with lines.
- To make it smooth, not only are the points joined (i.e. the equation meets at the points) but also the gradients.
- Equations (typically cubic: $ax^3+bx^2+cx+d$) describe curve between adjacent points, in this case the first and second differentials are equal at that point.
- Data values are assumed to be exact so requires accurate input data.

**Least squares fitting (curve fitting):**

- A fit is a way of using a function to represent all the data. It tells you something about your complete data set.
- One type of equation fitting is the least squares method.
- The differences between the data and the fit are _error_. This can be used to calculate the coefficient of determination $(R^2)$ which indicates the goodness of the fit.
- More suitable for most experiments.

## Spline fitting

Spline fitting can be used to either interpolate or extrapolate data. The latter however is needs careful consideration. 

### Interpolation using spline fitting
Spline fitting requires the data to be entered into the `CubicSpline()` function, which returns a function, where a number or array of numbers can be evaluated. If the function is required for future evaluations, it can be assigned to a variable `cs = CubicSpline(x, y)`. Otherwise you can directly evaluate the function `y1 = CubicSpline(x, y)(x1)`.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline

x = np.array([0,10,20,30,40,50])           # generate x data
y = np.array([5, 4.3, 4, 4.15,4.07,4.1])   # generate y data

x1 = np.linspace(5,15,50)                  # put 50 points between x=5 to x=15
cs = CubicSpline(x,y)                      # Creates spline function
y1 = cs(x1)                                # use x1 points to generate y1 points. 

plt.plot(x1,y1, 'bo', markersize = 3, label = 'Cubic Spline')
plt.plot(x,y, 'r*', markersize = 7, label='Original Data')
plt.legend()                               # adds legend on the basis of plot labels
plt.show()
plt.close()
```

## Extrapolation using spline fitting
The example above shows interpolation (between measured data points) but we can also use it to extrapolate (outside the data range). The accuracy should be very carefully considered.

```python
x2 = np.linspace(40,60,50)   # put 50 points between x=48 to x=60
y2 = cs(x2)                  # uses CubicSpline function from previous example to generate y2 points. 

plt.plot(x2,y2, 'bo', markersize = 3, label = 'Cubic Spline')
plt.plot(x,y, 'r*', markersize = 7, label='Original Data')
plt.legend()
plt.show()
plt.close()
```

Note: here that the data looks like its stabilising at the value of 4 but the extrapolation only takes the last two points and predicts what happens next. A limitation of extrapolation and spline fitting.

## Least squares fitting of data
Fitting using the least squares method is fairly straightforward in Python. To calculate the $R^2$ value, coefficient of determination, is a bit more involved, For simplicity we calculate it manually.

### A simple least squares fit
You can use the function polyfit to extract the prefactors of the order (degree) of fit you are using. For example
- Order 1: Equation is $y=Ax+b$
- Order 2: Equation is $y=Bx^2+Ax+C$ 
- Order 3. etc

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0,1.1, 0.1) # Run x from 0 to 1 in steps of 0.1
y = np.array([0.9,6.9,31.6,21.5,34.4,68.2,66.4,114.9,155.9,199.3,209.6])

p1 = np.polyfit(x, y, 1)   
# Order 1 fit of y= Ax + B 
# p1 stores all the prefactors as a vector
# the first element of p1 is A
# the second element of p1 is B
print(f'fit: y={p1[0]:.2f}x{p1[1]:+.2f}')

y1 = p1[0] * x + p1[1]
#alternatively using np.polyval
y1 = np.polyval(p1, x)

#we can now use this plot to graph out
plt.plot(x, y,'ro', label='Raw data')
plt.plot(x, y1, 'b-', linewidth=2, label='fit: y=ax+b')

plt.legend() # add a legend
plt.xlabel('x-values') 
plt.ylabel('y-values')
plt.show()
plt.close()
```

## Higher order least squares fitting
Least squares fits do not have to be linear. Higher order fits are created by changing the order (degree) in the `np.polyfit()` function. Note: the order of the coefficients is descending (highest order first, zeros order (constant) last). 

Note: The order (degree) has to be an integer, so if you want to fit a square root (order 0.5) you will have to think about rearrange the equation you are fitting.

```python
x = np.arange(0,1.1, 0.1) # Run x from 0 to 1 in steps of 0.1
y = np.array([0.9,6.9,31.6,21.5,34.4,68.2,66.4,114.9,155.9,199.3,209.6])

# second order fit y = Ax^2 + Bx + C
p1 = np.polyfit(x, y, 2)   

#p1[0] stores A, p1[1] stores B, p1[2] stores C
print(f'fit: y={p1[0]:.2f}x^2{p1[1]:+.2f}x{p1[2]:+.2f}')

#evaluates fit using np.polyval
y1 = np.polyval(p1, x)

#we can now use this plot to graph out
plt.plot(x, y,'ro', label='Raw data')
plt.plot(x, y1, 'b-', linewidth=2, label='fit: y=ax^2+bx+c')

plt.legend() # add a legend
plt.xlabel('x-values') 
plt.ylabel('y-values')
plt.show()
plt.close()
```

## Coefficient of determination (R-squared or $R^2$)
The coefficient of determination $(R^2)$ is **a number between 0 and 1 that measures how well a statistical model predicts an outcome** (goodness of fit). You can interpret the $R^2$ as the proportion of variation in the dependent variable that is predicted by the statistical model.
